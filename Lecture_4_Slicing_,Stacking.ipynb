# Stacking


**For combining two numpy array we have horizontal stacking and vertical stacking.**

**hstack()**

-Is used to horizontally stack arrays i.e. along axis=1

```python
# Customer ID , Nmae
c=np.array([[101,'Mira'],
            [102,'Abdul'],
            [103,'Andrea']])
# Customer ID, Purchase Amount ,Purchase Date
d=np.array([
    [101,250.50,'2023-08-01'],
    [102,150.00,'2023-08-02'],
    [103, 300.75,'2023-08-93']
])

np.hstack((c,d)) # horizontal stacking and the 2 numpy arrays are passed as a tuple
```

**vstack()**

```python
c=np.array([[101,'Mira'],
            [102,'Abdul'],
            [103,'Andrea']])
d=np.array([[104 ,'Venkat'],
            [105,'David'],
            [105,'Aryan']])
np.vstack((c,d)) # vertical stacking along axis=0
```

# Spliting

**A numpy array can be split into multiple sub-arrays using horizontal and vertical spliting**

**hsplit()**

-Split an array into multiple sub-arrays horizontally (column-wise) i.e. along axis=1

```python
transactions=np.array([
    [101,'Mohan',250.50,'2023-08-01'],
    [102,'Bob',150.00,'2023-08-02'],
    [103,'Fatima',300.75,'2023-08-01'],
    [104,'David',400.20,'2023-08-03'],
    [105,'Aryan',330.1,'2023-08-04']

])
transactions
```

```python
x,y=np.hsplit(transactions,[2]) # split along columns(axis=1) .Here [2]-->column index
x
```

```python
y
```

**vsplit()**

-Split an array into multiple sub-arrays vertically(row-wise) i.e. along axis=0

```python
transactions=np.array([
    [101,'Mohan',250.50,'2023-08-01'],
    [102,'Bob',150.00,'2023-08-02'],
    [103,'Fatima',300.75,'2023-08-01'],
    [104,'David',400.20,'2023-08-03'],
    [105,'Aryan',330.1,'2023-08-04']

])
transactions
```

```python
w,z=np.vsplit(transactions,[2]) # split from row 2
w
```

```python
z
```

# Masking

-Used for filtering values in numpy array.

-There are two ways to carry out boolean masking-

**1) Returning the result array**

**2) Returning a boolean array**

**Returning The Result Array**

-condition in indexing bracket [] of an array an it returns an array with required result

```python
# Method 1: Returning the result array

m=np.arange(15)
m
```

```python
m[m>5] # Here we pass a condition in indexing bracket [] of an array an it returns an array with required result
```

```python
m[m%2==0]
```

**Returning The Boolean Array**

-Returns a  boolean array of same size as the array it represents.

-The boolean array only contains value of either True or False

-The boolean array is called mask array or simple mask

```python
m1=np.arange(20)
m1
```

```python
mask=m1>8
print(mask)
```

```python
m1[mask] # the mask array can be passed in the index brackets of arr to return values that satisfy the condition.
```

```python
monthly_sales=np.array([30,33,35,28,42])
monthly_sales
```

```python
result=monthly_sales<32
result
```

```python
monthly_sales[result]  # 30 and 28 are the two sales values that are less than the threshold 32
```

# argmax()

-Returns the indices of the maximum values along an axis.

```python
monthly_sales=np.array([30,33,35,28,42])
monthly_sales
```

```python
print(np.max(monthly_sales))


```

```python
index=np.argmax(monthly_sales)
print(index)
```

```python
print(monthly_sales[index]) # To get the value back
```

For 2-D array

```python
import numpy as np

arr = np.array([[10, 20, 30],
                [40, 50, 60]])
arr
```

```python
print(np.max(arr))
```

```python
print(np.argmax(arr)) #  For a 2D array, it flattens the array by default and returns the index as if it's a 1D array.
```

```python
# To get the row index (axis=0)
print(np.argmax(arr,axis=0)) #from [[10, 20, 30],
                             #[40, 50, 60]] -->40 ,50,60 are max values along axis=0 and indexes of these three are [1 1 1]
```

```python
print(np.argmax(arr,axis=1)) #from [[10, 20, 30],
                             #[40, 50, 60]] --->30 and 60 are max values along axis=1 and hence has index [2 2]
```
